<h1>Algoritmo Horner</h1>

<p>Da base n a base 10</p>

<p>x = (((an−1 · 10 + an−2) · 10 + an−3) + . . .) · 10 + a0</p>

```python
# Horner bn to b10
def horner(number, base):

    output = 0

    length = len(str(number))#lenght 12345 = 5

    numberString = str(number)#"number"

    for i in range(0, length):
        #(((an−1 · 10 + an−2) · 10 + an−3) + . . .) · 10 + a0
        #
        output = output * base + int(numberString[i]) 

    return output

```

```python
output = horner(10010,2)

print("output: {} ".format(output))
```

```python
output = horner(1201,3)

print(f"output: {output}")
```

```python
output = horner( 1010111100010000,2)

print("output: " + str(output))
```
