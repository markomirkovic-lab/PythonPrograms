<p>Bracketing</p>

Condizione iniziale di partenza:
Funzione f(x)
(f(a)*f(b))<0 

poi, steps:		
(1) calcolo il valore di c		=(a+b)/2
(2) calcolo il valore delal f(x) nei vari punti		
(3) if f(a) * f(c)  < 0 --> prendo il primo sottointervallo, 
    ossia [a,c]		
    quindi a rimane a e b = c	
Else --> prendo il secondo sottointervallo, ossia [c,b]		
    quindi a=c e b rimane b	
POI RIPETO DA CAPO
