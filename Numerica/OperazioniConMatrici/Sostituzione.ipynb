<h1>Algoritmo di sostituzione</h1>
<p>• INPUT: sistema triangolare A, b (con A triangolare)</p>
<p>• OUTPUT: soluzione x tale che A · x = b</p>


```python
A = [[1,-1,2],
     [2,-1,-1],
     [1,1,1]]

b = [2,0,3]

"""
tmp = A[2][2]
x[2] = (b[2]-tmp)/A[2][2]
tmp = A[1][2]*x[2]
x[1] = (b[1]-tmp)/A[1][1]
tmp = A[0][2]*x[2] + A[0][1]*x[1]
x[0] = (b[0]-tmp)/A[0][0]

print(x[0])
print(x[1])
print(x[2])
"""
```

```python
import numpy



n = len(b)
x = numpy.array([0.0 for _ in range(n)])
for i in range(n-1,-1,-1):
    tmp = 0.0
    for j in range(n-1,i,-1):
        tmp += A[i][j]*x[j]
    x[i] = (b[i]-tmp)/A[i][i]
    print(x[i])
```
