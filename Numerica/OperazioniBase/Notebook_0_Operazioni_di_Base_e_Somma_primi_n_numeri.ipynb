```python
def somma_primi_n_numeri(n):
  somma = (n+1)*n/2
  return somma

def somma_primi_n_numeri_slow(n):
  somma = 0
  for contatore in range(n+1):
    somma = somma + contatore
  return somma

print(somma_primi_n_numeri(100))
print(somma_primi_n_numeri_slow(100))
```
