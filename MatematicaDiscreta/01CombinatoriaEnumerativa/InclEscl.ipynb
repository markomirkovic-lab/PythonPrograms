<p>Il principio di inclusione-esclusione</p>

```python
import time as tm

c = 0
t=tm.time()
for i in range(1,101):
    if i%2 != 0 and i%3 != 0:
        c=c+1
print(tm.time()-t)
print(c)
```

```python
c = 0
t2=tm.time()
for i in range(1,10000000+1):
    if i%2 != 0 and i%3 != 0:
        c=c+1
print(tm.time()-t2)
print(c)
```

```python
c = 0
t2=tm.time()
for i in range(1,511):
    if i%2 != 0 and i%3 != 0 and i%7!=0:
        c=c+1
print(tm.time()-t2)
print(c)
```
