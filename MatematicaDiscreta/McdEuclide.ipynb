```python
a = 1331
n = 1789

def bezuot(a,n):
    r1=[n,0]
    r2=[a,1]
    while r2[0] > 0:
        r3=[r1[0]%r2[0],r1[1]-r1[0]//r2[0]*r2[1]]
        r1=r2
        r2=r3
    return(r1[1]%n)
```

```python
print("inverso di ", a, " e' " , bezuot(1331,1789))

b = bezuot(a,n)

print("check: ")
print("a*b %n = ", (a*b)%n)

```

```python
inverseNumber = 0

def bezout_extended(a,n):
    r1 = [n,0,1,0]
    r2 = [a,0,0,1]
    print(r1)
    print(r2)
    while r2[0]>0:
        q=r1[0]//r2[0]
        r3 = [r1[0]%r2[0],q,r1[2]-q*r2[2],r1[3]-q*r2[3]]
        print(r3)
        r1=r2
        r2=r3
        #print(r3[0:])
        if(r3[0]==1):
            inverseNumber=(r3[3])
    return(r1[3]%n)

print("Inverse number: ", inverseNumber)
```
