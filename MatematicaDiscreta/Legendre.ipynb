```python
def mcd(a,n):
    if n==0:
        return(a)
    else:
        return(mcd(n,a%n))
```

```python
a=391
n=323
mcd_1 = mcd(a,n)
print('mcd: ', 2, ' mod ', n, ' = ', mcd_1)
```

```python
def pot(a,b,n):
    z=1
    for i in range(1,b+1):
        z=(z*a)%n
    return(z)
```

```python
a=14348
b=26704
n=29009
pot_1 = pot(a,b,n)
print('pot: ', pot_1)
```

```python
def bezuot_1(a,n):
    r1=[n,0]
    r2=[a,1]
    while r2[0] > 0:
        r3=[r1[0]%r2[0],r1[1]-r1[0]//r2[0]*r2[1]]
        r1=r2
        r2=r3
    return(r1[1]%n)
```

```python
a=1234567
n=7654321
b=bezuot_1(a,n)
print('Inverso di ', a, ' in Z(', n, ") = ", b)
```

```python
def bezuot(e,n):
    r1=[n,0]
    r2=[e,1]
    while r2[0] > 0:
        r3=[r1[0]%r2[0],r1[1]-r1[0]//r2[0]*r2[1]]
        r1=r2
        r2=r3
    return(r1[1]%n)
```

```python
def legendre(a,b,n):
    x=a
    y=b
    z=1
    while y > 0:
        if y%2==1:
            z = (z * x) % n#se dispari aggiorna z
        y=y//2
        x=(x*x)%n
    return(z)
```
